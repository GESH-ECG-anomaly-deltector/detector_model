```python
from tensorflow.keras.datasets import mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()

from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train.reshape(-1, x_train.shape[1]*x_train.shape[2])).reshape(-1, x_train.shape[1], x_train.shape[2])
x_test=sc.transform(x_test.reshape(-1, x_test.shape[1]*x_test.shape[2])).reshape(-1, x_test.shape[1], x_test.shape[2])
```

```python
model=keras.models.Sequential()

model.add( keras.layers.Conv2D(filters=50, kernel_size=5, activation="relu", padding="same", input_shape=[28, 28, 1]))
model.add( keras.layers.MaxPooling2D(2))

model.add( keras.layers.Conv2D(filters=100, kernel_size=5, activation="relu", padding="same"))
model.add( keras.layers.MaxPooling2D(2))

model.add( keras.layers.Flatten())

model.add( keras.layers.Dense(100, activation="relu" ))
model.add( keras.layers.Dropout(0.2))

model.add(keras.layers.Dense(10, activation="softmax"))
```

```python
model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

model.fit(
    x_train, y_train, 
    epochs=10, 
    validation_split=0.2
)
```

```python
input_data=x_train 

model=keras.models.Sequential()

input_=keras.layers.Input(shape=input_data.shape)
conv1=keras.layers.Conv2D()(input_)
conv2=keras.layers.Conv2D()(conv1)
conv3=keras.layers.Conv2D()(conv2)

max_pool=keras.layers.MaxPooling2D()(conv3)
conv4=keras.layers.Conv2D()(conv3)

conv5=keras.layers.Conv2D()(max_pool)
conv6=keras.layers.Conv2D()(conv4)
conv7=keras.layers.Conv2D()(conv4)
conv8=keras.layers.Conv2D()(conv4)
conv9=keras.layers.Conv2D()(conv3)

concat=keras.layers.Concatenate()(conv5, conv6, conv7, conv8)
batch_norm1=keras.layers.BatchNormalization()(concat)
batch_norm2=keras.layers.BatchNormalization()(conv9)

element_wise_addition=keras.layers.add()(batch_norm1, batch_norm2)

output=keras.activations.relu()(element_wise_addition)
```
