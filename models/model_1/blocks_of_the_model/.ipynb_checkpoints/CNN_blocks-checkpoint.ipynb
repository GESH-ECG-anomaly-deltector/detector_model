```python
import torch.nn as nn
```

```python
class conv_block(nn.Module):
    def __init__(self, filters, stride, padding, out_channels, in_channels):
        super().__init__()

        self.block=nn.sequential(
            nn.Conv1d(
                in_channels=in_channels, 
                out_channels=out_channels, 
                kernel_size=filters,
                stride=stride,
                padding=padding                
            )
        )


    def forward(self, x):
        return self.block(x)
```
