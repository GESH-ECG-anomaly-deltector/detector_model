```python
import tensorflow as tf
from tensorflow import keras
import sklearn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time
```

```python
from tensorflow.keras.datasets import mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()

from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train.reshape(-1, x_train.shape[1]*x_train.shape[2])).reshape(-1, x_train.shape[1], x_train.shape[2])
x_test=sc.transform(x_test.reshape(-1, x_test.shape[1]*x_test.shape[2])).reshape(-1, x_test.shape[1], x_test.shape[2])
```

```python
model=keras.models.Sequential()

model.add( keras.layers.Conv2D(filters=50, kernel_size=5, activation="relu", padding="same", input_shape=[28, 28, 1]))
model.add( keras.layers.MaxPooling2D(2))

model.add( keras.layers.Conv2D(filters=100, kernel_size=5, activation="relu", padding="same"))
model.add( keras.layers.MaxPooling2D(2))

model.add( keras.layers.Flatten())

model.add( keras.layers.Dense(100, activation="relu" ))
model.add( keras.layers.Dropout(0.2))

model.add(keras.layers.Dense(10, activation="softmax"))
```

```python
model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

model.fit(
    x_train, y_train, 
    epochs=10, 
    validation_split=0.2
)
```

```python
x_train=x_train.reshape(-1, 28, 28, 1)
x_test =x_test.reshape(-1, 28, 28, 1)
```

```python
#Hyperparameters:
#shape:   (filter, stride, padding, out_channels)
h_parameters=[
        [7,  1, 3,  128],
        [5,  1, 2,  256],
        [3,  1, 1,  128],
        [1,  1, 0,  32],
        [1,  1, 0,  32],
        [39, 1, 19, 32],
        [19, 1, 9,  32],
        [9,  1, 4,  32],
        [1,  1, 0,  128],    
    ]

#max_pooling hp(hyperparameter):
max_pool_hp=2



#functional APIs:


input_data=x_train
########################################################



input_=keras.layers.Input(shape=input_data.shape[1:])

conv_num=1
padding_1=keras.layers.ZeroPadding2D(h_parameters[0][2])(input_)
conv1=keras.layers.Conv2D(kernel_size=h_parameters[conv_num-1][0], strides=h_parameters[conv_num-1][1], padding="same", filters=h_parameters[conv_num-1][3], activation="relu")(padding_1)

conv_num=2
padding_2=keras.layers.ZeroPadding2D(h_parameters[1][2])(conv1)
conv2=keras.layers.Conv2D(kernel_size=h_parameters[conv_num-1][0], strides=h_parameters[conv_num-1][1], padding="same", filters=h_parameters[conv_num-1][3], activation="relu")(padding_2)

conv_num=3
padding_3=keras.layers.ZeroPadding2D(h_parameters[2][2])(conv2)
conv3=keras.layers.Conv2D(kernel_size=h_parameters[conv_num-1][0], strides=h_parameters[conv_num-1][1], padding="same", filters=h_parameters[conv_num-1][3], activation="relu")(padding_3)
########################################################



max_pool=keras.layers.MaxPooling2D(max_pool_hp)(conv3)

conv_num=4
padding_4=keras.layers.ZeroPadding2D(h_parameters[3][2])(conv3)
conv4=keras.layers.Conv2D(kernel_size=h_parameters[conv_num-1][0], strides=h_parameters[conv_num-1][1], padding="same", filters=h_parameters[conv_num-1][3], activation="relu")(padding_4)
########################################################



conv_num=5
padding_5=keras.layers.ZeroPadding2D(h_parameters[4][2])(max_pool)
conv5=keras.layers.Conv2D(kernel_size=h_parameters[conv_num-1][0], strides=h_parameters[conv_num-1][1], padding="same", filters=h_parameters[conv_num-1][3], activation="relu")(padding_5)

conv_num=6
padding_6=keras.layers.ZeroPadding2D(h_parameters[5][2])(conv4)
conv6=keras.layers.Conv2D(kernel_size=h_parameters[conv_num-1][0], strides=h_parameters[conv_num-1][1], padding="same", filters=h_parameters[conv_num-1][3], activation="relu")(padding_6)

conv_num=7
padding_7=keras.layers.ZeroPadding2D(h_parameters[6][2])(conv4)
conv7=keras.layers.Conv2D(kernel_size=h_parameters[conv_num-1][0], strides=h_parameters[conv_num-1][1], padding="same", filters=h_parameters[conv_num-1][3], activation="relu")(padding_7)

conv_num=8
padding_8=keras.layers.ZeroPadding2D(h_parameters[7][2])(conv4)
conv8=keras.layers.Conv2D(kernel_size=h_parameters[conv_num-1][0], strides=h_parameters[conv_num-1][1], padding="same", filters=h_parameters[conv_num-1][3], activation="relu")(padding_8)

conv_num=9
padding_9=keras.layers.ZeroPadding2D(h_parameters[8][2])(conv3)
conv9=keras.layers.Conv2D(kernel_size=h_parameters[conv_num-1][0], strides=h_parameters[conv_num-1][1], padding="same", filters=h_parameters[conv_num-1][3], activation="relu")(padding_9)

########################################################


# Your tensor shapes (Height, Width)
heights = [
    conv5.shape[1], 
    conv6.shape[1], 
    conv7.shape[1], 
    conv8.shape[1]
]

widths  = [
    conv5.shape[2],
    conv6.shape[2], 
    conv7.shape[2], 
    conv8.shape[2]
]

m_height = max(heights)
m_width  = max(widths) 

padded_5=keras.layers.ZeroPadding2D(padding=((0, -heights[0]+m_height), (0, -widths[0]+m_width))) (conv5)
padded_6=keras.layers.ZeroPadding2D(padding=((0, -heights[1]+m_height), (0, -widths[1]+m_width))) (conv6)
padded_7=keras.layers.ZeroPadding2D(padding=((0, -heights[2]+m_height), (0, -widths[2]+m_width))) (conv7)
padded_8=keras.layers.ZeroPadding2D(padding=((0, -heights[3]+m_height), (0, -widths[3]+m_width))) (conv8)

concat=keras.layers.Concatenate()([padded_5, padded_6, padded_7, padded_8])

batch_norm1=keras.layers.BatchNormalization()(concat)
batch_norm2=keras.layers.BatchNormalization()(conv9)
########################################################


# Your tensor shapes (Height, Width)
heights = [
    batch_norm1.shape[1], 
    batch_norm2.shape[1]
]

widths  = [
    batch_norm1.shape[2],
    batch_norm2.shape[2]
]

m_height = max(heights) 
m_width  = max(widths)   

padded_norm1=keras.layers.ZeroPadding2D(padding=((0, -heights[0]+m_height), (0, -widths[0]+m_width))) (batch_norm1)
padded_norm2=keras.layers.ZeroPadding2D(padding=((0, -heights[1]+m_height), (0, -widths[1]+m_width))) (batch_norm2)

element_wise_addition=keras.layers.Add()([padded_norm1, padded_norm2])
########################################################



relu=keras.layers.Activation("relu")(element_wise_addition)
########################################################



#Middle network:
middle_1=keras.layers.Flatten()(relu)

middle_2=keras.layers.Dense(100, activation="relu")(middle_1)
middle_3=keras.layers.Dropout(0.2)(middle_2)

#*******************************************************

output_=keras.layers.Dense(1, activation="softmax")(middle_3)
```

```python
model=keras.Model(inputs=[input_], outputs=[output_])
model.compile(
    loss="categorical_crossentropy", 
    optimizer="adam", 
    metrics=["accuracy"]
)
```
